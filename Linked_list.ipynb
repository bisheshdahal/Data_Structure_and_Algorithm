# Singly Linked List

```python

class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

    def __repr__(self):
        return str(self.data)


n1 = Node(55)
print(n1)
```

A `Node` is just a small box that holds one value (`data`) and a pointer (`next`) to whatever box comes after it. On its own it doesn't know about any list — it's the building block everything else is made of. Printing a node shows the value it's carrying.

```python

class SinglyLinkedList:
    def __init__(self):
        self.head = None
        self.tail = None
        self.count = 0

    def is_empty(self):
        return self.head is None

    def __len__(self):
        return self.count

    def __repr__(self):
        nodes = []
        current = self.head
        while current:
            nodes.append(str(current.data))
            current = current.next
        return " -> ".join(nodes) + " -> None"

    def insert_at_beginning(self, data):
        new_node = Node(data)
        self.count += 1

        if self.is_empty():
            self.head = new_node
            self.tail = new_node
        else:
            new_node.next = self.head
            self.head = new_node

    def insert_at_end(self, data):
        new_node = Node(data)
        self.count += 1

        if self.is_empty():
            self.head = new_node
            self.tail = new_node
        else:
            self.tail.next = new_node
            self.tail = new_node

    def insert_after_node(self, prev_node, data):
        if not prev_node:
            print("Error: previous node is None")
            return

        if prev_node == self.tail:
            self.insert_at_end(data)
            return

        new_node = Node(data)
        new_node.next = prev_node.next
        prev_node.next = new_node
        self.count += 1

    def delete_at_beginning(self):
        if self.is_empty():
            print("Error: the list is empty")
            return None

        self.count -= 1
        node_to_delete = self.head

        if self.head == self.tail:
            self.head = None
            self.tail = None
        else:
            self.head = self.head.next

        return node_to_delete.data

    def delete_at_end(self):
        if self.is_empty():
            print("Error: the list is empty")
            return None

        self.count -= 1

        if self.head == self.tail:
            node_data = self.head.data
            self.head = None
            self.tail = None
            return node_data

        prev = self.head
        while prev.next != self.tail:
            prev = prev.next

        node_data = self.tail.data
        prev.next = None
        self.tail = prev
        return node_data

    def delete_by_key(self, key):
        if self.is_empty():
            print("Error: the list is empty")
            return

        if self.head.data == key:
            self.delete_at_beginning()
            return

        prev = self.head
        current = self.head.next
        while current:
            if current.data == key:
                prev.next = current.next
                if current == self.tail:
                    self.tail = prev
                self.count -= 1
                return
            prev = current
            current = current.next

        print(f"Error: key '{key}' not found in the list.")

    def traverse(self):
        print("SLL", end=" ")
        current = self.head
        while current:
            print(current.data, end=" -> ")
            current = current.next
        print("None")

    def search(self, key):
        current = self.head
        while current:
            if current.data == key:
                return current
            current = current.next
        return None
```

This is the actual list. It only ever remembers two things directly — the `head` (first node) and `tail` (last node) — and every other node is reached by following `next` pointers from there.

- **Insert at beginning / end** just rewires a couple of pointers, so both are fast (O(1)) because we keep a `tail` reference.
- **Insert after a node** splices a new node in between two existing ones.
- **Delete at beginning / end / by key** removes a node and patches the pointers around it back together so the chain stays unbroken.
- **Traverse** walks from `head` to `None`, printing each value along the way.
- **Search** does the same walk but stops early if it finds the value you're looking for.

```python

sll = SinglyLinkedList()
sll.insert_at_beginning(7)
print(sll)

sll.insert_at_beginning(20)
print(sll)
```

Every time we insert at the beginning, the new value becomes the new front of the list, so `20` ends up ahead of `7`.

```python

sll.insert_at_end(13)
print(sll)
```

`13` is added after the current last node, so it becomes the new tail.

```python

sll.delete_at_beginning()
print(sll)
```

This removes whatever is currently at the front and shifts the next node up to take its place as `head`.

```python

sll.delete_at_end()
print(sll)
```

This walks the list until it finds the node just before the tail, then drops the tail and makes that node the new tail.

```python

sll.search(7)
```

Looks for `7` starting from the head and returns the node if it's still in the list (or `None` if it isn't).

# Doubly Linked List

```python

class DNode:
    def __init__(self, data):
        self.data = data
        self.next = None
        self.prev = None

    def __repr__(self):
        return str(self.data)
```

Same idea as before, but now each box also remembers the box **before** it (`prev`), not just the one after (`next`). That's what lets us walk the list in both directions.

```python

class DoublyLinkedList:
    def __init__(self):
        self.head = None
        self.tail = None
        self.count = 0

    def is_empty(self):
        return self.head is None

    def __len__(self):
        return self.count

    def __repr__(self):
        nodes = []
        current = self.head
        while current:
            nodes.append(str(current.data))
            current = current.next
        return "None <-> " + " <-> ".join(nodes) + " <-> None"

    def traverse_forward(self):
        print("DLL forward:", end=" ")
        current = self.head
        while current:
            print(current.data, end=" <-> ")
            current = current.next
        print("None")

    def traverse_backward(self):
        print("DLL backward:", end=" ")
        current = self.tail
        while current:
            print(current.data, end=" <-> ")
            current = current.prev
        print("None")

    def insert_at_beginning(self, data):
        new_node = DNode(data)
        self.count += 1

        if self.is_empty():
            self.head = new_node
            self.tail = new_node
            return

        new_node.next = self.head
        self.head.prev = new_node
        self.head = new_node

    def insert_at_end(self, data):
        new_node = DNode(data)
        self.count += 1

        if self.is_empty():
            self.head = new_node
            self.tail = new_node
            return

        self.tail.next = new_node
        new_node.prev = self.tail
        self.tail = new_node

    def insert_after_node(self, prev_node, data):
        if prev_node is None:
            print("Error: previous node is None")
            return

        if prev_node == self.tail:
            self.insert_at_end(data)
            return

        new_node = DNode(data)
        new_node.next = prev_node.next
        new_node.prev = prev_node

        prev_node.next.prev = new_node
        prev_node.next = new_node

        self.count += 1

    def delete_at_beginning(self):
        if self.is_empty():
            print("Error: list is empty")
            return None

        node_data = self.head.data
        self.count -= 1

        if self.head == self.tail:
            self.head = None
            self.tail = None
            return node_data

        self.head = self.head.next
        self.head.prev = None
        return node_data

    def delete_at_end(self):
        if self.is_empty():
            print("Error: list is empty")
            return None

        node_data = self.tail.data
        self.count -= 1

        if self.head == self.tail:
            self.head = None
            self.tail = None
            return node_data

        self.tail = self.tail.prev
        self.tail.next = None
        return node_data

    def delete_by_key(self, key):
        if self.is_empty():
            print("Error: list is empty")
            return

        if self.head.data == key:
            self.delete_at_beginning()
            return

        current = self.head
        while current:
            if current.data == key:
                if current == self.tail:
                    self.delete_at_end()
                    return
                current.prev.next = current.next
                current.next.prev = current.prev
                self.count -= 1
                return
            current = current.next

        print(f"Error: key '{key}' not found in list")
```

This is the doubly linked list, built on `DNode`. Because each node keeps a `prev` link as well as `next`, it can do a couple of things a singly linked list can't do cheaply:

- **Traverse backward** — walk from the tail back to the head.
- **Delete a node in O(1)** once you already have a reference to it, since you don't need to search for "the node before it" — it's right there in `prev`.

Insertion and deletion just have twice as many pointers to keep in sync (both `next` and `prev` on the neighboring nodes), but the overall logic mirrors the singly linked list.

```python

dll = DoublyLinkedList()
dll.insert_at_end(10)
dll.insert_at_end(20)
dll.insert_at_beginning(5)
print(dll)

dll.traverse_forward()
dll.traverse_backward()

dll.delete_at_beginning()
print(dll)

dll.delete_by_key(20)
print(dll)
```
