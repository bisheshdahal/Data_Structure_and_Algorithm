# Bubble Sort

```python

def bubble_sort(elements):
    size = len(elements)

    for j in range(size - 1):
        for i in range(size - 1):
            if elements[i] > elements[i + 1]:
                elements[i], elements[i + 1] = elements[i + 1], elements[i]


a = [12, 32, 42, 121, 12, 12, 3, 4, 5, 67]
bubble_sort(a)
print(a)

b = [12, 1, 1, 1, 1, 2, 2, 2, 3, 3, 1, 212, 1, 2, 1312, 23, 13, 2, 23, 12, 312, 3]
bubble_sort(b)
print(b)
```

Bubble sort works by repeatedly walking through the list and swapping any two neighbors that are in the wrong order. Each full pass "bubbles" the largest remaining value to the end, a bit like a bubble rising to the top of a glass — which is where the name comes from.

A couple of things worth noting:
- **Why `size - 1` passes?** After `n - 1` passes, the biggest `n - 1` values are already guaranteed to be in their final spots, so the very last value has nowhere else to go — the list is sorted.
- **Why `size - 1` comparisons per pass?** Each comparison looks at a pair `(i, i + 1)`, and the last valid pair in a list of length `n` starts at index `n - 2`, so the loop only needs to run `n - 1` times.
- **Why does `bubble_sort` return `None`?** It sorts the list **in place** — it rearranges the same list object you passed in rather than building and returning a new one — so `print(bubble_sort(a))` would just print `None`; you print `a` itself afterward instead.

It's simple to understand but not efficient: with two nested loops it runs in O(n²) time, which is fine for small lists but slow for large ones.
